```python
df = pd.read_csv('titanic.csv')
```

```python
df.head()
```

```python
features = ['Pclass', 'Age', 'Sex', 'Fare', 'SibSp', 'Parch']
target = 'Survived'
```

```python
data = df[features + [target]].copy()
```

```python
data.isnull().sum()
```

```python
data['Age'] = data['Age'].fillna(data['Age'].median())
```

```python
data['Sex'] = data['Sex'].map({'male': 0, 'female': 1})
```

```python
data.info()
data.isnull().sum()
data.head()
```

```python
corr_matrix = data.corr()
```

```python
corr_matrix_rounded = corr_matrix.round(2)
```

```python
print(corr_matrix_rounded)
```

```python
survived_corr = corr_matrix['Survived'].drop('Survived').sort_values(ascending=False)
```

```python
print(survived_corr)
```

```python
most_positive = survived_corr.idxmax()
```

```python
most_negative = survived_corr.idxmin()
```

```python
print(f"Наиболее сильная положительная корреляция: {most_positive} ({survived_corr.max()})")
```

```python
print(f"Наиболее сильная отрицательная корреляция: {most_negative} ({survived_corr.min()})")
```

```python
plt.figure(figsize=(10, 8))
```

```python
sns.heatmap(
    corr_matrix_rounded,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    vmin=-1, vmax=1
)
```

```python
plt.title('Корреляционная матрица признаков Titanic')
```
