```python
plt.figure(figsize=(12, 6))
plt.bar(genre_sales_sorted.index, genre_sales_sorted.values, color="steelblue")
plt.title("Мировые продажи по жанрам")
plt.xlabel("Жанр")
plt.ylabel("Продажи, млн")
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
region_corr = df_clean[
    ["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales"]
].corr()
```

```python
print(region_corr)
```

```python
plt.figure(figsize=(7, 5))
plt.imshow(region_corr, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Корреляция")
plt.xticks(range(len(region_corr.columns)), region_corr.columns)
plt.yticks(range(len(region_corr.columns)), region_corr.columns)
```

```python
for i in range(len(region_corr)):
    for j in range(len(region_corr)):
        plt.text(j, i, f"{region_corr.iloc[i, j]:.2f}",
                 ha="center", va="center", color="black")
```

```python
plt.title("Корреляция продаж по регионам")
plt.tight_layout()
plt.show()
```
